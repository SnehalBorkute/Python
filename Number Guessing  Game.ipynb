```python
import random

while True:

    print("Welcome to this interesting game...")
    print("1 - Easy level")
    print("2 - Hard level")

    level = int(input("Choose level: "))

    if level == 1:
        computer = random.randint(1, 10)
    elif level == 2:
        computer = random.randint(1, 50)
    else:
        print("Invalid level")
        continue

    guess = int(input("Guess the number: "))

    while guess != computer:

        if guess < computer:
            print("Too low")
        elif guess > computer:
            print("Too high")

        guess = int(input("Guess again: "))

    print("Correct!.. YOU WON")

    print("Choose from:")
    print("1 - Play Again")
    print("2 - Exit")

    user = int(input("Choose 1 or 2: "))

    if user == 1:
        continue

    elif user == 2:
        print("Thanks for playing!")
        break

    else:
        print("Invalid input")
        break
```
