```python
# generate random data and apply filters
import pandas as pd
import random
products=[]
for i in range(5):
    id=random.randint(1,10)
    pricing={"laptop":5000,"mobile":10000,"tv":12000,"induction":20000,"AC":65000}
    name=random.choice(["laptop","mobile","tv","induction","AC"])
    price=pricing[name]
    product={
        "productid":id,
        "productname":name,
        "price":price
    }
    products.append(product)

df=pd.DataFrame(products)
df.set_index("productid")
df[(df["productid"]%2==0) & (df["price"]>500)]
```
