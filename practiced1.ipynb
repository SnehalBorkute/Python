```python
# Amstrong number
import math
power = 0
n=153
temp=n
while(n>0):
    rem=n%10
    power+=int(math.pow(rem,3))
    
    n=int(n/10)
print(power)

if(power==temp):
    print("amstrong")
 
   

```

```python
#reverse a number
n=1234
rem=0;
while(n>0):
    num=n%10
    rem=rem*10+num
  
    
    n=int(n/10)
print(rem)
```

```python
#Count the frequency of each element in a list.
#USING DICT
a = [1, 2, 2, 3, 1, 2]
d={}
count=0
for i in  a:
    if i not in d:
        d[i]=count+1
    else:
        d[i]=d[i]+1

    
print(d)


```

```python
#Find common elements in two lists
a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]
c=[]
for i in a:
    if(i in b):
        c.append(i)
print(c)

```
