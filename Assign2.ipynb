```python
import numpy as np
```

```python
#2d array slicing

_a=np.array([  [10,30,50],[43,55,65]  ])
_a
```

```python
print(_a[0]) # 0th row
print(_a[1])

print(_a[0:2]) # both 0,1st row
```

```python
print(_a[0,1]) #0th row 1st column
print(_a[1,0]) 
```

```python
print(_a[0,0:3]) #0th row and values from 0,1,2 columns

print(_a[1,0:2]) #0th row and values from 0,1 columns
```

```python
print(_a[0:2,0:3]) #(row,column) that why 0,1 row and 0,1,2 column
```
