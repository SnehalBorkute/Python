```python
# Develop a Python-based **News Aggregator** that allows users to select a news categorysuch as Technology, Sports, Business, or Science andfetch the 
# latest related news using **NewsAPI**.The application should display the news title, source, description, and URL in a readable format 
# and allow users to search for another category or exit.



import requests

API_KEY = "YOUR_API_KEY"
topics = {
    "1": "technology",
    "2": "sports",
    "3": "business",
    "4": "science",
    "5": "health",
    "6": "entertainment",
    "7": "artificial intelligence"
}

while True:

    print("\n" + "=" * 70)
    print("                         NEWS APP")
    print("=" * 70)

    print("\nChoose a topic:\n")
    print("1. Technology")
    print("2. Sports")
    print("3. Business")
    print("4. Science")
    print("5. Health")
    print("6. Entertainment")
    print("7. Artificial Intelligence")
    print("8. Exit")

    choice = input("\nEnter your choice (1-8): ")

    if choice == "8":
        print("\nThank you for using the News App!")
        break

    if choice not in topics:
        print("\nInvalid choice! Please select 1-8.")
        continue

    topic = topics[choice]

    url = "https://newsapi.org/v2/everything"

    params = {
        "q": topic,
        "apiKey": API_KEY,
        "language": "en",
        "sortBy": "publishedAt",
        "pageSize": 10
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:

        data = response.json()
        articles = data["articles"]

        print("\n" + "=" * 70)
        print(f"                    {topic.upper()} NEWS")
        print("=" * 70)

        for i, article in enumerate(articles, start=1):

            print(f"\n{i}. {article['title']}")

            print("\nSource:")
            print(f"   {article['source']['name']}")

            print("\nDescription:")
            print(f"   {article['description']}")

            print("\nRead Full News:")
            print(f"   {article['url']}")

            print("\n" + "-" * 70)

    else:
        print("\nError:", response.status_code)
        print(response.text)

    input("\nPress Enter to choose another topic...")
```
