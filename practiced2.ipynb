```python
#missing number
a = [1, 2, 4, 5]
for i in range(1,6):
    if i not in a:
        print(i)
```

```python
#using list comprehension
a = [1, 2, 4, 5]
missing = [i for i in range(1,6) if i not in a]
print(missing)
```

```python
# count a letters in word and store in dict
s = "apple"
d={}
count=0
for l in s:
    if l not in d:
        d[l]=count+1
    else:
        d[l]+=1
print(d)
```

```python
#dict comprehension
s = "apple"
d={l:s.count(l) for l in s}
print(d)
```

```python
# Create a dictionary containing only even numbers greater than 10,
# where each number is the key and its square is the value.

a = [12, 7, 20, 15, 8, 25, 30]
[i**2 for i in a if i%2==0 and i>10]
```
